```python
station_coordinates = pd.DataFrame({
    "station": [
        "Station A",
        "Station B",
        "Station C",
        "Station D",
        "Station E"
    ],
    "lat": [
        40.735,
        40.751,
        40.742,
        40.728,
        40.760
    ],
    "lng": [
        -73.991,
        -73.977,
        -73.985,
        -73.970,
        -73.995
    ]
})

station_coordinates
```

```python
start_df =trip_demo.merge(station_coordinates, left_on="start_station", right_on="station", how="left")
start_df.rename(columns={"lat": "start_lat", "lng": "start_lng"}, inplace=True)
start_df.drop(columns=["station"], inplace=True)
start_df.head()
```

```python
end_df =trip_demo.merge(station_coordinates, left_on="end_station", right_on="station", how="left")
end_df.rename(columns={"lat": "end_lat", "lng": "end_lng"}, inplace=True)
end_df.drop(columns=["station"], inplace=True)
end_df.head()
```

```python
final_df = pd.merge(start_df[["ride_id", "start_station", "started_at", "start_lat", "start_lng"]], end_df[["ride_id", "end_station", "ended_at", "end_lat", "end_lng", "member_casual"]], on="ride_id", how="inner")
final_df.head()
```

```python
map_center = [
    pd.concat([final_df
    ["start_lat"], final_df
    ["end_lat"]]).mean(),
    pd.concat([final_df
    ["start_lng"], final_df
    ["end_lng"]]).mean()
]

map_center
```

```python
final_df["duration_min"] = (
    final_df["ended_at"] - final_df["started_at"]
).dt.total_seconds() / 60

final_df[[
    "ride_id",
    "start_station",
    "end_station",
    "started_at",
    "ended_at",
    "duration_min"
]].head()
```

```python
final_df["date"] = final_df["started_at"].dt.date
final_df["hour"] = final_df["started_at"].dt.hour
final_df["day_name"] = final_df["started_at"].dt.day_name()
final_df["month_name"] = final_df["started_at"].dt.month_name()

final_df.head()
```

## Folium

```python
m = folium.Map(
    location=[40.74, -73.99],
    zoom_start=13
)

for _, row in station_coordinates.iterrows():
    folium.Marker(
        location=[row["lat"], row["lng"]],
        popup=row["station"],
        tooltip=row["station"]
    ).add_to(m)

m
```

```python
m = folium.Map(
    location=[40.74, -73.99],
    zoom_start=13
)

start_point = [final_df.loc[0, "start_lat"], final_df.loc[0, "start_lng"]]
end_point = [final_df.loc[0, "end_lat"], final_df.loc[0, "end_lng"]]

folium.Marker(start_point, popup="Start").add_to(m)
folium.Marker(end_point, popup="End").add_to(m)

folium.PolyLine(
    locations=[start_point, end_point],
    weight=3,
    opacity=0.6,
    color="green"
).add_to(m)

m
```

## Bounding Box

```python
min_lat = min(final_df["start_lat"].min(), final_df["end_lat"].min())
max_lat = max(final_df["start_lat"].max(), final_df["end_lat"].max())

min_lng = min(final_df["start_lng"].min(), final_df["end_lng"].min())
max_lng = max(final_df["start_lng"].max(), final_df["end_lng"].max())

bounding_box = pd.DataFrame({
    "metric": ["min_lat", "max_lat", "min_lng", "max_lng"],
    "value": [min_lat, max_lat, min_lng, max_lng]
})

bounding_box
```

```python
bbox_map = folium.Map(
    location=map_center,
    zoom_start=13,
    tiles="CartoDB positron"
)

for _, row in station_coordinates.iterrows():
    folium.CircleMarker(
        location=[row["lat"], row["lng"]],
        radius=6,
        popup=row["station"],
        tooltip=row["station"],
        fill=True
    ).add_to(bbox_map)

bbox_coordinates = [
    [min_lat, min_lng],
    [min_lat, max_lng],
    [max_lat, max_lng],
    [max_lat, min_lng],
    [min_lat, min_lng]
]

folium.PolyLine(
    locations=bbox_coordinates,
    weight=4,
    opacity=0.8,
    popup="Bounding Box"
).add_to(bbox_map)

bbox_map
```
